#For Dataset GSE16463

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
import pandas as pd
import gzip

file_path = "/content/drive/MyDrive/Dengue_Infection/GSE16463/GSE16463_series_matrix.txt.gz"

with gzip.open(file_path, 'rt') as f:
    lines = f.readlines()
```

```python
start = None
end = None

for i, line in enumerate(lines):
    if "!series_matrix_table_begin" in line:
        start = i + 1
    if "!series_matrix_table_end" in line:
        end = i
        break

data_lines = lines[start:end]
```

```python
from io import StringIO

data_str = "".join(data_lines)
df = pd.read_csv(StringIO(data_str), sep="\t")
```

```python
df.rename(columns={df.columns[0]: "ID"}, inplace=True)
```

```python
df = df.dropna()
```

```python
gene_col = None

for col in df.columns:
    if "gene" in col.lower() or "symbol" in col.lower():
        gene_col = col
        break

print("Gene column:", gene_col)
```

```python
if gene_col is not None:
    df = df.rename(columns={gene_col: "Gene.symbol"})

    df = df.dropna(subset=["Gene.symbol"])

    # Split multiple genes in same row (if exists)
    df["Gene.symbol"] = df["Gene.symbol"].astype(str).str.split("///")
    df = df.explode("Gene.symbol")

    # Remove duplicates (keep mean expression)
    df_gene = df.groupby("Gene.symbol").mean(numeric_only=True)
else:
    df_gene = df.set_index("ID")
```

```python
output_path = "/content/drive/MyDrive/Dengue_Infection/GSE16463/GSE16463_RMA_GeneLevel.csv"

df_gene.to_csv(output_path)

print("Saved:", output_path)
print("Shape:", df_gene.shape)
```
