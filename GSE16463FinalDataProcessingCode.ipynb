```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd

datasets = [
    "GSE28405",
    "GSE51808",
    "GSE84331",
    "GSE18090",
    "GSE28985"
]

base_dir = "/content/drive/MyDrive/Dengue_Infection"

for ds in datasets:

    print(f"\nProcessing {ds}")

    # File paths
    expr_file = f"{base_dir}/{ds}/{ds}_RMA_GeneLevel.csv"
    deg_file = f"{base_dir}/{ds}/{ds}.top.table.tsv"

    # Load files
    expr = pd.read_csv(expr_file)
    deg = pd.read_csv(deg_file, sep="\t")

    # Rename first column of expression matrix
    expr.rename(columns={expr.columns[0]: "Probe.ID"}, inplace=True)

    # Remove NA Gene.symbol
    deg = deg.dropna(subset=["Gene.symbol"])

    # Rename ID column
    deg.rename(columns={"ID": "Probe.ID"}, inplace=True)

    # Merge
    merged_df = pd.merge(
        expr,
        deg[['Probe.ID', 'Gene.symbol', 'adj.P.Val', 'P.Value', 'logFC']],
        on='Probe.ID',
        how='left'
    )

    print("Merged shape:", merged_df.shape)

    # -------------------
    # Condition 1
    # -------------------
    DEG_P025 = merged_df[
        merged_df["P.Value"] < 0.25
    ].copy()

    DEG_P025.drop(
        columns=["adj.P.Val", "P.Value", "logFC"],
        inplace=True
    )

    DEG_P025.to_csv(
        f"{base_dir}/{ds}/{ds}_DEG_P025.csv",
        index=False
    )

    print("P<0.25 shape:", DEG_P025.shape)

    # -------------------
    # Condition 2
    # -------------------
    DEG_2Condition = merged_df[
        (merged_df["adj.P.Val"] < 0.1) &
        (merged_df["logFC"] > 0.5)
    ].copy()

    DEG_2Condition.drop(
        columns=["adj.P.Val", "P.Value", "logFC"],
        inplace=True
    )

    DEG_2Condition.to_csv(
        f"{base_dir}/{ds}/{ds}_DEG_2Condition.csv",
        index=False
    )

    print("adj.P.Val<0.1 & logFC>0.5 shape:", DEG_2Condition.shape)

print("\nAll datasets completed successfully.")
```

#**Combet and Batch Effect using batch variance**

```python
import pandas as pd
from functools import reduce

# =========================
# Dataset groups
# =========================
train_datasets = [
    "GSE51808",
    "GSE84331",
    "GSE18090",
    "GSE16463",
    "GSE28985"
]

test_dataset = "GSE28405"

base_path = "/content/drive/MyDrive/Dengue_Infection"

# =========================
# Function to load dataset
# =========================
def load_deg(dataset, condition="2Condition"):
    path = f"{base_path}/{dataset}/{dataset}_DEG_{condition}.csv"
    df = pd.read_csv(path)

    # ensure Gene.symbol exists
    df = df.dropna(subset=["Gene.symbol"])
    return df


# =========================
# Load TRAIN datasets
# =========================
train_list = []

for ds in train_datasets:
    df = load_deg(ds, "2Condition")
    df = df[["Gene.symbol"]] + df.drop(columns=["Gene.symbol"], errors="ignore")
    train_list.append(df)

# =========================
# Merge TRAIN datasets (inner join on Gene.symbol)
# =========================
train_merged = reduce(
    lambda left, right: pd.merge(left, right, on="Gene.symbol", how="inner"),
    train_list
)

print("✅ Train merged shape:", train_merged.shape)


# =========================
# Load TEST dataset
# =========================
test_df = load_deg(test_dataset, "2Condition")

# =========================
# Merge TEST with TRAIN genes
# =========================
final_merged = pd.merge(
    test_df,
    train_merged,
    on="Gene.symbol",
    how="inner",
    suffixes=("_TEST", "_TRAIN")
)

print("✅ Final merged (Train + Test) shape:", final_merged.shape)


# =========================
# Save raw merged dataset
# =========================
final_merged.to_csv(
    f"{base_path}/FINAL_MERGED_ALL_DATASETS.csv",
    index=False
)


# =========================
# Gene symbol as index
# =========================
df = final_merged.set_index("Gene.symbol")

df.to_csv(
    f"{base_path}/FINAL_INDEXED_DATASET.csv"
)

print("✅ Saved successfully")
```

```python
import pandas as pd

# =========================
# FUNCTION (MUST RUN FIRST)
# =========================
def get_gene_set(path):
    df = pd.read_csv(path)

    # clean columns
    df.columns = df.columns.str.strip()

    # gene column detect
    if "Gene.symbol" in df.columns:
        gene_col = "Gene.symbol"
    else:
        gene_col = df.columns[-1]   # fallback (your case)

    genes = (
        df[gene_col]
        .astype(str)
        .str.upper()
        .str.strip()
        .dropna()
        .unique()
    )

    return set(genes)


# =========================
# COMMON GENE PIPELINE
# =========================
base_path = "/content/drive/MyDrive/Dengue_Infection"

datasets = [
    "GSE28405",
    "GSE51808",
    "GSE84331",
    #"GSE18090",
    #"GSE16463",
    "GSE28985"
]

gene_sets = []

for ds in datasets:
    path = f"{base_path}/{ds}/{ds}_DEG_2Condition.csv"
    genes = get_gene_set(path)

    print(ds, "genes:", len(genes))
    gene_sets.append(genes)

common_genes = set.intersection(*gene_sets)

print("\n✅ FINAL COMMON GENES:", len(common_genes))
```

```python
!pip install upsetplot

import matplotlib.pyplot as plt
from upsetplot import from_contents, UpSet

# =========================
# Data
# =========================
gene_dict = dict(zip(datasets, gene_sets))

# Convert
upset_data = from_contents(gene_dict)

# =========================
# Colors for each dataset
# =========================
color_map = {
    datasets[0]: "#FF9999",  # pink
    datasets[1]: "#66B3FF",  # blue
    datasets[2]: "#99FF99",  # green
    datasets[3]: "#FFCC99",  # orange
    datasets[4]: "#C2C2F0",  # purple
    datasets[5]: "#FFD700"   # gold
}

# =========================
# Plot
# =========================
plt.figure(figsize=(14,6))

up = UpSet(
    upset_data,
    subset_size='count',
    show_counts=True,
    sort_by='cardinality'
)

up.plot()

# =========================
# FIXED COLOR APPLICATION
# =========================
ax = plt.gcf().axes[0]

bars = ax.patches

# assign colors cyclically per dataset (correct way)
for i, bar in enumerate(bars):
    dataset_index = i % len(datasets)
    bar.set_facecolor(color_map[datasets[dataset_index]])
    bar.set_edgecolor("black")

plt.suptitle(
    "UpSet Plot: 6 Dengue Datasets (Color-coded)",
    fontsize=14,
    fontweight='bold'
)

plt.show()
```
